```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

```python
img=cv2.imread('../DATASETS/images/she.jpg')
img
```

```python
q_width,q_height=img.shape[1]//4,img.shape[0]//4
```

```python
T = np.float32([[1, 0, q_width], [0, 1, q_height]])
```

```python
T
```

```python
trans_img = cv2.warpAffine(img, T, (600,600))
```

```python
cv2.imshow('Translated Image', trans_img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
mask = (trans_img == 0)
trans_img[mask] = 255
cv2.imshow('Translated with White Background', trans_img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
img_90 = cv2.rotate(img, cv2.ROTATE_90_CLOCKWISE)
img_90a =cv2.rotate(img, cv2.ROTATE_90_COUNTERCLOCKWISE)
img_180 = cv2.rotate(img, cv2.ROTATE_180) 
```

```python
cv2.imshow('90 Clockwise', img_90)
cv2.imshow('90 Anti-Clockwise', img_90a)
cv2.imshow('180 Degrees', img_180)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

### Calculate the center coordinates of the image

```python

center = img.shape[1] // 2, img.shape[0] // 2
```

```python
center
```

```python
angle=30
R=cv2.getRotationMatrix2D(center,angle,1.0)
```

### Applying the rotation and displaying the output

```python
img_rot = cv2.warpAffine(img, R, (400, 400))
cv2.imshow('Rotated Image', img_rot)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# [30]: Custom rotation with hardcoded center
angle = -65
R = cv2.getRotationMatrix2D((300, 300), angle, 1.0)
img_rot = cv2.warpAffine(img, R, (400,400))
cv2.imshow('Rotated Image', img_rot)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
