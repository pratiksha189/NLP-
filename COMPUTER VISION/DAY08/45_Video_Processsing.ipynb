* EXTRACT ALL THE FRAMES AND FRAME SAND STORE IN FOLDER /nature

```python
vid = cv2.VideoCapture('../DATASETS/images/nature.mp4')
count=1
while True:
    ret,frame = vid.read()
    if ret == False:
        break
    cv2.imwrite(f'AllImages/image{count}.jpg',frame)
    count+=1
cv2.destroyAllWindows()
print('all Frames saved Successfully!')
```

```python
time = '00:00:42.00'
hrs, mins, secs = time.split(':')
hrs, mins, secs
fps = vid.get(cv2.CAP_PROP_FPS)
tframe = int(hrs) * 3600 * fps + int(mins) * 60 * fps + float(secs) * fps
tframe
```

```python
vid = cv2.VideoCapture(0)

frame_width = int(vid.get(cv2.CAP_PROP_FRAME_WIDTH))
frame_height = int(vid.get(cv2.CAP_PROP_FRAME_HEIGHT))
fourcc = cv2.VideoWriter_fourcc(*'XVID')
out = cv2.VideoWriter('output.avi', fourcc, 30.0, (frame_width, frame_height))

if not vid.isOpened():
    print("Error: Could not open webcam.")
else:
    while True:
        ret, frame = vid.read()
        out.write(frame)
        cv2.imshow('My Video', frame)
        if cv2.waitKey(5) == ord('q'):
            break

vid.release()
out.release()
cv2.destroyAllWindows()
```
