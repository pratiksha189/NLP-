```python
import torch
import matplotlib.pyplot as plt
import cv2
from PIL import Image
```

```python
model=torch.hub.load('ultralytics/yolov5','yolov5s')
```

```python
# define the predict function
def predict(image):
  image=Image.open(image)
  results=model(image)
  print('Detected image',results.names)
  results.show()
```

```python
predict('cat.jpg')
```

```python
predict('person.jpg')
```

```python
predict('jungle.jpg')
```
