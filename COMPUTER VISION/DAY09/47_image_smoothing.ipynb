```python


img = cv2.imread('../DATASETS/images/dog.jpg')
```

```python

kernel = np.zeros([3,3]) + 1/9
```

```python
blurred_img = cv2.filter2D(img, -1, kernel)
```

```python

cv2.imshow('Blurred Image', blurred_img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python

kernel = np.zeros([9,9]) + 1/81
blurred_img = cv2.filter2D(img, -1, kernel)
img_new = np.hstack([img, blurred_img])
cv2.imshow('Blurred Image', img_new)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

### Gaussian kernel

```python
kernel_size=7
kernel=cv2.getGaussianKernel(kernel_size,1.23)
```

```python
kernel
```

```python
gauss_image = cv2.filter2D(img, -1, kernel)
img_new = np.hstack([img, gauss_image])
cv2.imshow('Blurred Image', img_new)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

### median blur

```python
img_gray=cv2.imread('../DATASETS/images/dog.jpg',0)
img_gray.shape
```

```python
img_med=cv2.medianBlur(img_gray,7)
```

```python
plt.figure(figsize=(16,9))
plt.subplot(121)
plt.imshow(img_gray, cmap = 'gray')
plt.subplot(122)
plt.imshow(img_med, cmap = 'gray')
```

```python
# 1. Make a copy of the image so you don't overwrite the original
blurred_roi_img = img_gray.copy()

# 2. Define your region coordinates (ymin, ymax, xmin, xmax)
# Let's say we want to blur a 150x150 pixel box in the upper middle area
ymin, ymax = 100, 250
xmin, xmax = 100, 250

# 3. Extract the region, blur it, and assign it right back to the copy
roi = blurred_roi_img[ymin:ymax, xmin:xmax]
blurred_roi_img[ymin:ymax, xmin:xmax] = cv2.medianBlur(roi, 5)

# 4. Display the result
plt.figure(figsize=(10, 6))
plt.imshow(blurred_roi_img, cmap='gray')
plt.show()
```
