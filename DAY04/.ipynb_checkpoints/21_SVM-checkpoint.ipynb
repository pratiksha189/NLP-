##### data preprocessing
###### data cleaning 


```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import LancasterStemmer, WordNetLemmatizer
from nltk import pos_tag

```

```python
# Sample text data
sent = 'Hello friends! How are you? I like Python Programming.'

# 1. Tokenize the data
tokens1 = word_tokenize(sent)
print("Original Tokens:", tokens1)

# 2. Remove punctuation and numbers
tokens2 = [token for token in tokens1 if token.isalpha()]
print("Cleaned Tokens:", tokens2)
```

```python

tokens3 = [token.lower() for token in tokens2 
           if token.lower() not in stopwords.words('english')]
tokens3
```

```python
# Initialize the Lancaster Stemmer
ls = LancasterStemmer()

# Apply stemming to the clean tokens
tokens4 = [ls.stem(token) for token in tokens3]

# View the result
tokens4
```

```python
from nltk.stem import PorterStemmer

# apply stemming
ps = PorterStemmer()
tokens4 = [ps.stem(token) for token in tokens3]

# View the result
tokens4
```

```python
def clean_text(sent):
    # tokenize the data
    tokens1 = word_tokenize(sent)
    
    # remove punctuation and numbers
    tokens2 = [token for token in tokens1 if token.isalpha()]
    
    # remove stopwords
    tokens3 = [token.lower() for token in tokens2
               if token.lower() not in stopwords.words('english')]
               
    tokens4 = [ps.stem(token) for token in tokens3]
    return tokens4

clean_text(sent)
```

```python
sent1='It was an unexpected rain during the play on ground '
```

```python
clean_text(sent1)
```

#### vectorization 

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer(analyzer=clean_text)
```

```python
x_new=tfidf.fit_transform(x)
```

```python
x_new.shape
```

```python
tfidf.get_feature_names_out()
```

```python
len(tfidf.get_feature_names_out())
```

```python
x_new.toarray()
```

#### cross validation

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(x_new,y,random_state=0)
```

```python
x_train.shape
```

```python
x_test.shape
```

```python
x_train, x_test, y_train, y_test = train_test_split(x_new, y, test_size=0.2, random_state=0)
```

```python
from sklearn.svm import SVC
```

```python
svc_lin = SVC(kernel='linear') 
```

```python
svc_lin.fit(x_train, y_train)
```

### performance evaluation


```python
from sklearn.metrics import accuracy_score,classification_report
from sklearn.metrics import ConfusionMatrixDisplay
```

```python
y_pred = svc_lin.predict(x_test)
```

```python
ConfusionMatrixDisplay.from_predictions(y_test,y_test)
```

```python
print(classification_report(y_test, y_pred))
```

```python
accuracy_score(y_test,y_pred)
```

#### using rbf kernel

```python
svc_rbf = SVC(kernel='rbf')
svc_rbf.fit(x_train, y_train)
y_pred = svc_rbf.predict(x_test)

print(accuracy_score(y_test, y_pred))
```

```python
print(classification_report(y_test, y_pred))
```

#### using polynomial kernel

```python

svc_poly = SVC(kernel='poly', degree=3)
svc_poly.fit(x_train, y_train)
y_pred = svc_poly.predict(x_test)

print(accuracy_score(y_test, y_pred))
```

```python
print(classification_report(y_test, y_pred))
```

```python
svc_sigmoid = SVC(kernel='sigmoid')
svc_sigmoid.fit(x_train, y_train)
y_pred = svc_sigmoid.predict(x_test)

print(accuracy_score(y_test, y_pred))
```

```python
print(classification_report(y_test, y_pred))
```
