```python
sent = "They told that their ages are 78, 24, 31 and 29 respectively."
# FIND AVERAGES OF AGES MNENTIONED IN THE SENTENCE.
print(sent.split())
```

```python
age_add = []
for i in sent.split():
    if i.isdigit():
        age_add.append(int(i))
print(sum(age_add)/len(age_add))
```

```python
ages=[int(i) for i in sent.split() if i.isdigit()]
ages
sum(ages)/len(ages)
```

```python
# find percentage of punctutions in the sentence.
punctuations = [i for i in sent if i in '.,!?;:']
punctuations
len(punctuations)
len(sent)   
percentage=len(punctuations)/len(sent)*100
percentage
```

```python
count_punct = 0
for word in word_tokenize(sent):
    if word in string.punctuation:
        count_punct += 1
count_punct
```

```python
count_punct / len(word_tokenize(sent)) * 100
```

```python
## remove punctuations from the sentence.
l = "hello friends! HOW ARE YOU?\n WELCOME TO WORLD OF PYTHON PROGRAMMING."
new_str=""
for char in l:
    if char not in string.punctuation:
        new_str += char
print(new_str)
```

```python
# 2. Now call the tokenize method
tokens = tk.tokenize(sent)
print(tokens)
```

## CUSTOM TOKENIZER

```python
import re
```

```python
def custom_tokenizer(text):
    return re.split(r"[.,:?!\s]+",text)
text="This is some text with punctuation > Let's tokenize it.Is it okk?"
```

```python
custom_tokenizer(text)
```

```python

class CommaTokenizer: 
    def custom_tokenizer(self,text):
        return re.split(r"[,\s]+",text)


    
```

```python
tk = CommaTokenizer()
```

```python
fruits = "apple, orange, banana, cherry, pineapple."
```

```python
tk.tokenize(fruits)
```
