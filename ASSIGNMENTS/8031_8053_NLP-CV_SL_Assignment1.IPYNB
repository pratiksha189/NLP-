```python
import re
import time
import spacy

nlp = spacy.load("en_core_web_sm")

messy_text = """
The grand event happened on 2024-10-15 in New York City. Another milestone was met on 12/05/2025 in London. 
Some random text here and there. Paris was lovely during the meeting on 02-18-2026.
"""

print("--- 1. LEXICAL ANALYSIS & PATTERN MATCHING ---")

start_time_regex = time.perf_counter()

date_pattern = r'(\d{4}-\d{2}-\d{2}|\d{2}/\d{2}/\d{4}|\d{2}-\d{2}-\d{4})'
location_pattern = r'\b[A-Z][a-z]+(?:\s+[A-Z][a-z]+)*\b'

regex_dates = re.findall(date_pattern, messy_text)
raw_loc_matches = re.findall(location_pattern, messy_text)
regex_locations = [loc for loc in raw_loc_matches if not re.match(r'\d', loc) and loc != "The"]

end_time_regex = time.perf_counter()
regex_duration = end_time_regex - start_time_regex

print(f"[Regex] Execution Time: {regex_duration:.6f} seconds")
print(f"[Regex] Extracted Dates: {regex_dates}")
print(f"[Regex] Extracted Locations (Rough Guessing): {regex_locations}\n")

start_time_spacy = time.perf_counter()

doc = nlp(messy_text)
spacy_dates = [ent.text for ent in doc.ents if ent.label_ == "DATE"]
spacy_locations = [ent.text for ent in doc.ents if ent.label_ in ["GPE", "LOC"]]

end_time_spacy = time.perf_counter()
spacy_duration = end_time_spacy - start_time_spacy

print(f"[spaCy] Execution Time: {spacy_duration:.6f} seconds")
print(f"[spaCy] Extracted Dates: {spacy_dates}")
print(f"[spaCy] Extracted Locations: {spacy_locations}")
```
