#### Fetch the Wikipedia page for the National Flag of India

```python

url = "https://en.wikipedia.org/wiki/National_flag_of_India"
headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36"
}
response = requests.get(url, headers=headers)
soup = BeautifulSoup(response.text, 'html.parser')
```

#### Extract raw text lines using a basic for loop

```python
text_lines = []
for p in soup.find_all('p'):
    if p.get_text().strip():
        text_lines.append(p.get_text().strip())
```

#### Combine lines into one string and break down into single words

```python

full_text = "\n".join(text_lines)
print(full_text)
all_words = full_text.split()
print(all_words)
```

#### 1. Count total number of Words

```python
total_words = len(all_words)
print(f"1. Total number of words: {total_words}")
```

####  2. Count percentage of punctuation marks




```python
punctuation_chars = 0
for char in full_text:
    if char in string.punctuation:
        punctuation_chars += 1

total_chars = len(full_text)
if total_chars > 0:
    punctuation_percentage = (punctuation_chars / total_chars) * 100
else:
    punctuation_percentage = 0
print(f"2. Percentage of punctuation marks: {punctuation_percentage:.2f}%")
```
