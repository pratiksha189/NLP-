```python
import nltk
```

```python
import random
```

```python
responses = {
    'greeting': ['Hello!', 'Hi there!', 'Hey!', 'Hiya!'],
    'goodbye': ['Good bye!', 'See you later!', 'Take care!', 'Bye!'],
    'thanks': ['You are welcome.', 'Glad to help.', 'Anytime.', 'Happy to assist!'],
    'default':['Sorry I didnt understand','can you tell again?',
               'I am not sure.']
}
```

```python
# Simple set of questions and responses or patterns
patterns = {
    'hi' : 'greeting',
    'hello' : 'greeting',
    'hey' : 'greeting',
    'bye' : 'goodbye',
    'thanks':'thanks',
    'thank you': 'thanks'
}
```

```python
# match the user 
def respond(user_input):
    for word in nltk.tokenize.word_tokenize(user_input.lower()):
        if word in patterns:
            intent = patterns[word]
            return random.choice(responses[intent])
    return random.choice(responses['default'])
```

```python
respond('hi')
```

```python
# # Match key words
def respond(user_input):
    for word in nltk.word_tokenize(user_input.lower()):
        if word in patterns:
            intent = patterns[word]
            return random.choice(responses[intent])
    return random.choice(responses['default'])

```

```python

# chatbot()
def chat():
    print("Chatbot: How can I help you today? (Type 'exit' to quit)")
    while True:
        user_input = input("You: ")
        if user_input.lower() == 'exit':
            break
        response = respond(user_input)
        print('Chatbot: ',response)
```

```python
chat()
```
