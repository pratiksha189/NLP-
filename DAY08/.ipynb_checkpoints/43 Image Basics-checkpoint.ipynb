```python
chess = np.zeros([8,8])
for i in range(8):
    for j in range(8):
        if (i+j) % 2 == 0:
            chess[i,j] = 1
```

```python
plt.imshow(chess,cmap='gray')
```

```python
# Orange - 255 , 165, 0
# Green - 0 , 255 , 0
# White - 255 , 0, 0

import numpy as np
import matplotlib.pyplot as plt

# 1. Initialize a 300x400 RGB image with zeros (Height=300, Width=400, Channels=3)
# Using uint8 data type for standard pixel values (0-255)
flag = np.zeros([300, 400, 3], dtype=np.uint8)

# 2. Assign the top band (Rows 0 to 100) -> Saffron / Orange [R=255, G=153, B=51]
flag[0:100, :, 0] = 255  # Red channel
flag[0:100, :, 1] = 153  # Green channel
flag[0:100, :, 2] = 51   # Blue channel

# 3. Assign the middle band (Rows 100 to 200) -> White [R=255, G=255, B=255]
flag[100:200, :, :] = 255 # Sets all three color channels to max brightness

# 4. Assign the bottom band (Rows 200 to 300) -> India Green [R=19, G=136, B=8]
flag[200:300, :, 0] = 19   # Red channel
flag[200:300, :, 1] = 136  # Green channel
flag[200:300, :, 2] = 8    # Blue channel

# 5. Display the generated image array using Matplotlib
plt.imshow(flag)
```
