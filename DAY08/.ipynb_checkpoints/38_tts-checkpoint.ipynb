```python
import pyttsx3

# Step 1: Initialize the text-to-speech engine
engine = pyttsx3.init()
voices = engine.getProperty('voices')

# Step 2: Set the voice to Microsoft Zira (Female)
# Index 1 is standard for Zira on Windows setups
engine.setProperty('voice', voices[1].id)

# Step 3: Lower the speech rate (Default is 200)
# Dropping it to 135 adds a slow-groove rap/song cadence
engine.setProperty('rate', 135)

# Step 4: Phonetically rewritten lyrics so an English voice engine 
# successfully mimics the Hindi/Punjabi pronunciation
singing_lyrics = """
ब्लू आइज़ हिप्नोटाइज़ तेरी करदी ऐ मैनु
आई स्वर छोटी ड्रेस में बॉम्ब लगदी मैनु
ग्लॉसी लिप्स, उफ़ ये ट्रिक्स, बेबी लगदी ऐ किलर
ओ ये, ओ ये, कत्ल करे तेरा बॉम्ब फिगर

[वर्श १]
मुंडे मरदे तेरे ते, तू लक्क नु हिलावे
जदों निकलदी ऐ बाहर, तू अग्ग लगावे
तेरा नखरा ऐ वखरा, तू सब नु सतावे
दिल सच्चा ऐ मेरा, तू क्यों आज़मावे?

[ब्रिज]
घर पता कर लिया तेरे घर दा
मैं रहंदा हुण डर-डर के
ओ मुंडे तेरे पीछे पए गए
रख लिया दिल हथेली धर के

[कोरस]
ब्लू आइज़ हिप्नोटाइज़ तेरी करदी ऐ मैनु
आई स्वर छोटी ड्रेस में बॉम्ब लगदी मैनु
"""

# Step 5: Execute the track
print("🎤 Zira is now performing 'Blue Eyes'...")
engine.say(singing_lyrics)
engine.runAndWait()
```

```python
from gtts import gTTS
from playsound import playsound

# --- Example 1: Marathi Language Speech Synthesis ---
# Create the object of TTS
tts_mr = gTTS(text='पायथॉन प्रोग्रॅमिंगमध्ये तुमचे स्वागत आहे.', lang='mr')

# Save and play the Marathi audio
tts_mr.save('audio_marathi.mp3')
playsound('audio_marathi.mp3')


# --- Example 2: English Language (UK Accent) Speech Synthesis ---
# Create the object of TTS with a UK top-level domain accent
tts_en = gTTS(text='Welcome to Python Programming.', lang='en', tld='co.uk')

# Save and play the English audio
tts_en.save('audio_english.mp3')
playsound('audio_english.mp3')
```
