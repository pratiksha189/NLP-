```python
from nltk.tokenize import word_tokenize
from nltk  import pos_tag,RegexpParser
```

```python
grammer="NP:{<DT><JJ><NN>}"
```

```python
sent = "It was a beautifull game by him. An amazing play was witnessed"
```

```python
parser = RegexpParser(grammer)
```

```python
tree = parser.parse(pos_tag(word_tokenize(sent)))
tree
```

```python
grammer = r'NP:{<DT>?<JJ>*<NN>+}'
# <DT>? Zero or one Determiner
# <DT>* zero or more adjectives
# <DT>+ One or more nouns
```

```python
sent = "It was a beautifull game by him. An amazing play was witnessed.They won the game."
parser = RegexpParser(grammer)
```

```python
tree = parser.parse(pos_tag(word_tokenize(sent)))
tree
```
