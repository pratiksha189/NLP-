```python
import pyiwn
```

```python
iwn = pyiwn.IndoWordNet(lang=pyiwn.Language.MARATHI)
```

```python
word = iwn.synsets('धरती')
word
```

```python
## explore the specific synsets
```

```python
word[0].synset_id()
```

```python
# meaning

word[0].gloss()
```

```python
# example in sentence
word[0].examples()
```

```python
word[0].pos()
```

```python
# synonyms
word[0].lemma_names()
```

```python
# hyponyms
hyponyms=iwn.synset_relation(word[0],pyiwn.SynsetRelations.HYPONYMY)
```

```python
for h in hyponyms:
    print(h.lemma()[0])
```

```python
word=iwn.synsets('आंबा')
hyponyms=iwn.synset_relation(word[0],pyiwn.SynsetRelations.HYPONYMY)
for h in hyponyms:
    print(h.lemma_names()[0])
```
