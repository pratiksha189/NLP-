```python
int_to_char=dict(enumerate(alphabets))
int_to_char
```

```python
char_to_int=dict((v,i) for i,v in enumerate( alphabets))
char_to_int
```

#### data preparation


```python
seq_length = 1
x = []
y = []

for i in range(len(alphabets) - seq_length):
    seq_in = [alphabets[i]]
    seq_out = alphabets[i+1]
    print(seq_in, '-->', seq_out)
    x += [[char_to_int[seq_in[0]]]]
    y += [char_to_int[seq_out]]
```

```python
import numpy as np
```

```python
x=np.reshape(x,(25,1,1))
```

```python
x_scaled=x/25
```

```python
from keras.utils import to_categorical
```

```python
y_new=to_categorical(y)
```

```python
y_new.shape
```

### built the model

```python
from keras.layers import SimpleRNN, Input, Dense
from keras.models import Sequential
model = Sequential()

model.add(Input((1, 1)))
model.add(Dense(32, activation='relu'))
model.add(SimpleRNN(32))
model.add(Dense(26, activation='softmax'))


model.summary()
```

```python
# [33] Compile the model
model.compile(loss='categorical_crossentropy',
              optimizer='adam', metrics=['accuracy'])

# [*] Train the model
model.fit(x_scaled, y_new, batch_size=1, epochs=500)
```

```python
model.fit(x_scaled, y_new, batch_size=1, epochs=1000)
```

```python
model.evaluate(x_scaled,y_new,batch_size=1)
```

```python

new_char = 'K'
new_char = char_to_int[new_char]
new_char = np.reshape(new_char, (1, 1, 1))
new_char = new_char / 25
pred = model.predict(new_char)
int_to_char[pred.argmax()]

```

```python
# [53] Prepare the data for a sequence length of 3
seq_length = 3
x = []
y = []

for i in range(len(alphabets) - seq_length):
    seq_in = alphabets[i:i+seq_length]
    seq_out = alphabets[i+seq_length]
    print(seq_in, '-->', seq_out)
    x.append([char_to_int[char] for char in seq_in])
    y += [char_to_int[seq_out]]
```

```python
x_new = np.reshape(x, (23, seq_length, 1))

y_new = to_categorical(y)

x_new.shape, y_new.shape

x_new = x_new / 25

model = Sequential()

model.add(Input((3, 1)))
model.add(Dense(32, activation='relu'))
model.add(SimpleRNN(32))
model.add(Dense(26, activation='softmax'))

model.compile(loss='categorical_crossentropy',
              optimizer='adam', metrics=['accuracy'])
```

```python
model.fit(x_new, y_new, batch_size=1, epochs=1000)
```

```python
new = 'JKL'
new_char = [char_to_int[char] for char in new]
new_char = np.array(new_char) / 25
new_char = np.reshape(new_char, (1, seq_length, 1))
pred = model.predict(new_char, verbose=0)
int_to_char[pred.argmax()]
```

```python
new = 'PQR'
new_char = [char_to_int[char] for char in new]
new_char = np.array(new_char) / 25
new_char = np.reshape(new_char, (1, seq_length, 1))
pred = model.predict(new_char, verbose=0)
int_to_char[pred.argmax()]
```
