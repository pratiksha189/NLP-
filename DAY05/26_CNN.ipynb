```python
def process_docs(directory, is_train):
    documents = []
    for filename in os.listdir(directory):
        # Skip test files (names starting with 'cv9') when training
        if is_train and filename.startswith('cv9'):
            continue
        # Skip training files (names NOT starting with 'cv9') when testing
        if not is_train and not filename.startswith('cv9'):
            continue
            
        path = directory + '/' + filename
        docs = load_doc(path)
        tokens = clean_doc(docs)
        documents.append(tokens)
        
    return documents
```

```python
def load_data(is_train):
    neg = process_docs('../DATASETS/txt_sentoken-20260603T032612Z-3-001/txt_sentoken/neg', is_train)
    pos = process_docs('../DATASETS/txt_sentoken-20260603T032612Z-3-001/txt_sentoken/pos', is_train)
    docs = neg + pos
    labels = [0 for i in range(len(neg))] + [1 for i in range(len(pos))]
    return docs, labels
```

```python
import numpy as np

class SimpleTokenizer:
    def __init__(self):
        self.word_index = {}
        
    def fit_on_texts(self, lines):
        # Flatten the list of lists into unique words
        vocab = set()
        for line in lines:
            vocab.update(line)
        # Build the word index mapping (starting from index 1)
        self.word_index = {word: i+1 for i, word in enumerate(sorted(vocab))}
        
    def texts_to_matrix(self, lines, mode='binary'):
        # Create an empty matrix of shape (number of documents, vocabulary size + 1)
        matrix = np.zeros((len(lines), len(self.word_index) + 1))
        
        for i, line in enumerate(lines):
            for word in line:
                if word in self.word_index:
                    idx = self.word_index[word]
                    if mode == 'binary':
                        matrix[i, idx] = 1
        return matrix

# Keeping the helper function intact
def create_tokenizer(lines):
    tokenizer = SimpleTokenizer()
    tokenizer.fit_on_texts(lines)
    return tokenizer
```

```python
from tensorflow.keras.preprocessing.sequence import pad_sequences

def encode_docs(tokenizer, max_length, docs):
    # This will now work perfectly with the Keras Tokenizer
    encoded = tokenizer.texts_to_sequences(docs)
    padded = pad_sequences(encoded, maxlen=max_length, padding='post')
    return padded
```

```python
# Encode the training and testing datasets using the encoder function
x_train = encode_docs(tokenizer, max_length, train_data)
x_test = encode_docs(tokenizer, max_length, test_data)
```

```python
from keras.models import Sequential
from keras.layers import Input, Dense, Conv1D, MaxPool1D, Embedding, Flatten
from keras.utils import to_categorical, plot_model
```

```python
from keras.models import Sequential
from keras.layers import Embedding, Conv1D, MaxPool1D, Flatten, Dense

# Calculate total vocabulary size from your tokenizer (adding 1 for padding token 0)
vocab_size = len(tokenizer.word_index) + 1

model = Sequential()

# Corrected Embedding Layer configuration
model.add(Embedding(input_dim=vocab_size, output_dim=128, input_length=max_length))

model.add(Conv1D(filters=16, kernel_size=5, activation='relu'))
model.add(MaxPool1D(pool_size=2))
model.add(Flatten())
model.add(Dense(16, activation='relu'))
# Add your final output layer depending on your task (e.g., binary classification)
model.add(Dense(1, activation='sigmoid'))
```

```python
model.fit(x_train, np.array(train_labels), batch_size=5, epochs=10)
```

```python
from tensorflow.keras.utils import plot_model

# Generate a visual graph of your sequential architecture
plot_model(model, to_file='model_architecture.png', show_shapes=True, show_layer_names=True)
```
