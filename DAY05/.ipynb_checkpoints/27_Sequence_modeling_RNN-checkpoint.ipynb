```python
alphabets='ABCDEFGHIJKLMNOPQRSTUVWXYZ'
```

```python
len(alphabets)
```

```python
list(enumerate(alphabets))
```

```python
int_to_char=dict(enumerate(alphabets))
int_to_char
```

```python
char_to_int=dict((v,i) for i,v in enumerate( alphabets))
char_to_int
```

#### data preparation


```python
seq_length = 1
x = []
y = []

for i in range(len(alphabets) - seq_length):
    seq_in = [alphabets[i]]
    seq_out = alphabets[i+1]
    print(seq_in, '-->', seq_out)
    x += [[char_to_int[seq_in[0]]]]
    y += [char_to_int[seq_out]]
```

```python
x
```

```python
y
```

```python
import numpy as np
```

```python
x=np.reshape(x,(25,1,1))
```

```python
x
```

```python
x_scaled=x/25
```

```python
x_scaled
```

```python
from keras.utils import to_categorical
```

```python
y_new=to_categorical(y)
```

```python
y_new.shape
```

### built the model

```python
from keras.layers import SimpleRNN, Input, Dense
from keras.models import Sequential
model = Sequential()

model.add(Input((1, 1)))
model.add(Dense(32, activation='relu'))
model.add(SimpleRNN(32))
model.add(Dense(26, activation='softmax'))


model.summary()
```

```python
# [33] Compile the model
model.compile(loss='categorical_crossentropy',
              optimizer='adam', metrics=['accuracy'])

# [*] Train the model
model.fit(x_scaled, y_new, batch_size=1, epochs=500)
```

```python
model.fit(x_scaled, y_new, batch_size=1, epochs=1000)
```

```python
model.evaluate(x_scaled,y_new,batch_size=1)
```

```python

new_char = 'K'

char_num = char_to_int[new_char]


scaled_num = char_num / 25.0

import numpy as np
input_data = np.reshape(scaled_num, (1, 1, 1))

prediction = model.predict(input_data)

```

```python
import numpy as np
input_data = np.reshape(scaled_num, (1, 1, 1))

prediction = model.predict(input_data)
```

```python

predicted_index = np.argmax(prediction)

predicted_char = int_to_char[predicted_index]

print(f"Input: {new_char} --> Predicted Next Character: {predicted_char}")
```
