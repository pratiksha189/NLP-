```python
import nltk
```

```python
#find the average of their ages mentioned in this sentence

sent = "They told that their ages are 24 31 and 29 respectively"
ages = [int(w) for w in sent.split() if w.isdigit()]

print("Average age: ",sum(ages)/ len(ages))
```

```python
#import the tokenizer function
from nltk.tokenize import word_tokenize, sent_tokenize
```

```python
sent = '''Hello friends!
How are you?
Welcome to the world of Python Programming.'''
```

```python
sents = sent_tokenize(sent)
print(sents)
```

```python
words = word_tokenize(sent)
print(words)
```

```python
import string
```

```python
string.punctuation
```

```python
#percentage of punctuation in the sentence
punc = [c for c in words if c in string.punctuation]
print(punc)
percentage = len(punc) * 100 /len(words) 
print(percentage)
```

### Custom Tokenizer

```python
import re
```

```python
def custom_tokenizer(text):
    return re.split(r"[,.:;?!\s]+", text)
```

```python
text = "This is some text with punctuations > Let's toknize it. Is it ok?"
custom_tokenizer(text)
```

```python
class CommaTokenizer:
    def tokenize(self, text):
        return re.split(r"[,\s]+", text)
```

```python
text = "Happy, You are happy"
ctk = CommaTokenizer()
ctk.tokenize(text)
```
