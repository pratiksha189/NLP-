```python
from nltk.tokenize import word_tokenize
from nltk import pos_tag
```

```python
sent = "A quick brown fox jumps over the lazy dog."
```

```python
tags = pos_tag(word_tokenize(sent))
```

```python
print(tags)
```

```python
paragraph = '''Indore  ISO: Indaura, Hindi is the largest city of the Indian state of Madhya Pradesh where it serves as the capital and the administrative headquarters of the eponymous district and division. Modern-day Indore was established on the banks of the Kanh and Saraswati rivers and traces its roots to its 16th-century founding as a trading hub between the Deccan Plateau and Delhi.[11][12] Following the conquest of Malwa by Peshwa Baji Rao I, the city came under the control of the Maratha Empire until its annexation by the British East India Company.[13][14]Prior to Indian independence, the Kingdom of Indore was a nineteen Gun Salute Princely State of the British Raj with the Holkars as rulers.[15]'''
```

```python
paratags = pos_tag(word_tokenize(paragraph))
paratags
```

```python
for tup in paratags:
    if tup[1] == "JJ":
        print(tup)
```

```python
def isReal(num):
    try:
        float(num)
        return True
    except ValueError:
        return False
```

```python
clean = [word for word in word_tokenize(paragraph) if word.isalpha() or isReal(word)]
```

```python
print(clean)
```

```python
print(pos_tag(clean))
```

```python
pune = '''Pune (Marathi: Puṇē, pronounced [ˈpuɳe] ⓘ), previously spelled in English as Poona (the official name until 1978),[16][17] is a city in the state of Maharashtra in the Deccan Plateau in Western India. It is the administrative headquarters of the Pune district, and of Pune division. In terms of the total amount of land under its jurisdiction, Pune is the largest city in Maharashtra by area, with a geographical area of 516.18km2,[18] though by population it comes in a distant second to Mumbai. According to the 2011 Census of India, Pune has 7.2 million residents in the metropolitan region, making it the seventh-most populous metropolitan area in India.[19] The city of Pune is part of Pune Metropolitan Region.[20] Pune is one of the largest IT hubs in India.[21][22] It is also one of the most important automobile and manufacturing hubs of India. Pune is often referred to as the "Oxford of the East" because of its educational institutions.[23][24][25] It has been ranked "the most liveable city in India" several times.[26][27]'''
```

```python
pune
```

```python
clean_pune = [word for word in word_tokenize(pune) if word.isalpha() or isReal(word)]
```

```python
pos_pune = pos_tag(clean_pune)
```

```python
#printing the adjectives - 
for word,tag in pos_pune:
    if tag.startswith("JJ"):
        print(word)
```

```python
pos_pune
```

```python
#printing connected adjective and noun pair 
pair_list = []
for tup_in in range(len(pos_pune) - 1):
    if pos_pune[tup_in][1].startswith("JJ") and pos_pune[tup_in + 1][1].startswith("NN"):
        pair_list.append([pos_pune[tup_in][0], pos_pune[tup_in + 1][0]])

pair_list
```
